                                                        THEORITICAL QUESTIONS.

1. What is Python, and why is it popular?  
Ans. Python is a high-level, interpreted programming language known for its simplicity and readability. It’s popular because it has a clean syntax, supports multiple programming paradigms, has huge community support, and comes with tons of useful libraries for things like data science, web development, and automation.

2. What is an interpreter in Python?  
Ans. The Python interpreter is a program that executes Python code line by line. It converts the human-readable Python code into machine-understandable instructions during runtime, which makes debugging and testing easier.

3.What are pre-defined keywords in Python?  
Ans. Pre-defined keywords are special reserved words that have specific meanings in Python. For example: if, else, for, while, True, False, None, and, or, etc. We can’t use these words as variable names because Python uses them for its own syntax rules.

4.Can keywords be used as variable names?  
Ans. No, keywords cannot be used as variable names. For example, writing if = 10 will cause a syntax error because if is a reserved keyword.

5️. What is mutability in Python?  
Ans.  Mutability means whether an object’s value can be changed after it’s created. If it can be changed, it’s mutable; if not, it’s immutable. For example, lists are mutable, but tuples and strings are immutable.

6. Why are lists mutable, but tuples are immutable?  
Ans. Lists are designed to store dynamic collections of items that can be modified, so Python allows changes to their elements. Tuples, on the other hand, are meant to be fixed collections, so once created, their elements cannot be changed—this makes them more memory-efficient and faster.

7. What is the difference between “==” and “is” operators in Python?  
Ans. The == operator checks whether the values of two objects are equal, while the is operator checks whether they both refer to the same memory location.

8️. What are logical operators in Python?  
Ans. Logical operators are used to combine conditional statements. The main ones are:

and → returns True if both conditions are true

or → returns True if at least one condition is true

not → reverses the logical result.

9️. What is type casting in Python?  
Ans. Type casting is the process of converting one data type into another. For example, converting an integer to a string or vice versa.

10. What is the difference between implicit and explicit type casting?  
Ans. Implicit type casting is done automatically by Python (no manual code needed), while explicit type casting is done manually by the programmer using functions like int(), float(), or str().
Example:

 -Implicit
x = 5
y = 2.5
z = x + y  # Python automatically converts int to float

 -Explicit
a = int(3.8)  # manually convert float to int


11. What is the purpose of conditional statements in Python?  
Ans. Conditional statements allow the program to make decisions based on certain conditions. They help control the flow of the program using if, elif, and else statements.

️12. How does the elif statement work?  
Ans . The elif statement checks another condition if the previous if condition is false. It stands for “else if.”
Example:

x = 10
if x > 20:
    print("Greater than 20")
elif x > 5:
    print("Greater than 5")
else:
    print("5 or less")


Here, since x > 5, the elif block runs.

13 . What is the difference between for and while loops?  
Ans. A for loop runs for a specific number of times or over a sequence (like a list or range).
A while loop runs until a certain condition becomes false.

14. Describe a scenario where a while loop is more suitable than a for loop?  
ans. A while loop is better when we don’t know beforehand how many times we’ll need to loop.
For example, repeatedly asking a user for input until they type “exit”.

```python
# 5. Write a function to demonstrate the behavior of mutable and immutable arguments
def modify_list(my_list):
    my_list.append(10)  # list is mutable

def modify_number(num):
    num += 5  # integer is immutable
    print("Inside function (num):", num)

numbers = [1, 2, 3]
n = 7

modify_list(numbers)
modify_number(n)

print("After function call (list):", numbers)
print("After function call (num):", n)
```

```python
# 8. Create a list and tuple in Python, and demonstrate how attempting to change an element works differently for each
# List (Mutable)
my_list = [1, 2, 3]
my_list[1] = 99
print("Modified List:", my_list)

# Tuple (Immutable)
my_tuple = (1, 2, 3)
try:
    my_tuple[1] = 99
except TypeError:
    print("Error: Tuples are immutable and cannot be changed!")
```

```python
# 13. Write a Python program to find the sum of all even numbers between 1 and 50
sum_even = 0
for i in range(1, 51):
    if i % 2 == 0:
        sum_even += i
print("Sum of even numbers from 1 to 50:", sum_even)
```

```python
# 14. Write a program to reverse a string using a while loop
text = input("Enter a string: ")
reversed_text = ""
i = len(text) - 1

while i >= 0:
    reversed_text += text[i]
    i -= 1

print("Reversed string:", reversed_text)
```

```python
# 15. Write a Python program to calculate the factorial of a number provided by the user using a while loop
num = int(input("Enter a number: "))
factorial = 1
i = 2

while i <= num:
    factorial *= i
    i += 1

print("Factorial of", num, "is:", factorial)
```
