1)What is Simple Linear Regression?

Simple Linear Regression is a statistical method used to model the relationship between one independent variable (X) and one dependent variable (Y) by fitting a straight line to the data.

The model is:

Y = mX + c

Where:
Y = Dependent variable
X = Independent variable
m = Slope
c = Intercept


2)- What are the key assumptions of Simple Linear Regression?

The main assumptions are:

Linearity – Relationship between X and Y is linear
Independence – Observations are independent of each other
Homoscedasticity – Constant variance of errors across all values of X
Normality of errors – Residuals are normally distributed
No outliers – No extreme influential data points


3- What does the coefficient m represent in the equation Y = mX + c?

The coefficient m represents the slope of the regression line.

It shows:
The change in Y for a one-unit increase in X
The direction and strength of the relationship between X and Y


4- What does the intercept c represent in the equation Y = mX + c?

The intercept c represents:
The value of Y when X = 0

It is where the regression line crosses the Y-axis and provides a baseline value of Y.


5- How do we calculate the slope m in Simple Linear Regression?

The slope m is calculated as:

m = [ Σ (X - X̄)(Y - Ȳ) ] / [ Σ (X - X̄)² ]

It is the ratio of:
Covariance of X and Y
To the variance of X

This formula ensures the best linear fit to the data.


6 -  What is the purpose of the least squares method in Simple Linear Regression?

The least squares method finds the best-fit line by:

Minimizing the sum of squared residuals (errors)

This ensures that:
The total prediction error is as small as possible
The regression line fits the data as closely as possible


7 -How is the coefficient of determination (R²) interpreted in Simple Linear Regression?

R² represents:
The proportion of variance in Y explained by X

Example:
If R² = 0.8, it means:
80% of Y’s variation is explained by X
20% of the variation is due to other factors or noise

8- What is Multiple Linear Regression?

Multiple Linear Regression models the relationship between:

One dependent variable (Y)
And two or more independent variables (X1, X2, …)

Equation:

Y = b0 + b1X1 + b2X2 + ... + bnXn


9- What is the main difference between Simple and Multiple Linear Regression?

Simple Linear Regression:
One independent variable
Y = mX + c

Multiple Linear Regression:
Two or more independent variables
Y = b0 + b1X1 + b2X2 + ...


10- What are the key assumptions of Multiple Linear Regression?

Linearity
Independence of errors
Homoscedasticity
Normality of residuals
No multicollinearity
No influential outliers


11- What is heteroscedasticity, and how does it affect the results of a Multiple Linear Regression model?

Heteroscedasticity means:
Error variance is not constant across observations

Effects:
Biased standard errors
Incorrect hypothesis tests
Unreliable confidence intervals
Reduced model reliability


12- How can you improve a Multiple Linear Regression model with high multicollinearity?

Ways to fix multicollinearity:

Remove highly correlated features
Use Variance Inflation Factor (VIF)
Use Ridge or Lasso regression
Apply Principal Component Analysis (PCA)
Combine correlated variables


13- What are some common techniques for transforming categorical variables for use in regression models?

Common techniques:

One-Hot Encoding
Label Encoding
Dummy Variables
Target Encoding
Binary Encoding


14- What is the role of interaction terms in Multiple Linear Regression?

Interaction terms model:
The combined effect of two variables on Y

Example:

Y = b1X1 + b2X2 + b3(X1 × X2)

They capture:
Dependency between predictors
Situations where the effect of one variable depends on another


15- How can the interpretation of intercept differ between Simple and Multiple Linear Regression?

In Simple Linear Regression:
Intercept = Y when X = 0

In Multiple Linear Regression:
Intercept = Y when all independent variables = 0
This may be unrealistic and less meaningful


16- What is the significance of the slope in regression analysis, and how does it affect predictions?

The slope shows:
Strength of the relationship
Direction (positive or negative)

It affects predictions by:
Determining how much Y changes for a one-unit change in X


17- How does the intercept in a regression model provide context for the relationship between variables?

The intercept gives:
A baseline value of Y

It helps:
Understand the starting point of the relationship
Provide context for predictions


18- What are the limitations of using R² as a sole measure of model performance?

Limitations:

Increases automatically with more variables
Does not detect overfitting
Does not show prediction accuracy
Does not indicate causation


19- How would you interpret a large standard error for a regression coefficient?

A large standard error means:
The coefficient estimate is unreliable

It suggests:
High variance in data
Multicollinearity
Small sample size


20- How can heteroscedasticity be identified in residual plots, and why is it important to address it?

Identified by:
Funnel or cone shape in residual vs fitted value plot

Important because it:
Invalidates statistical tests
Affects confidence intervals
Reduces reliability of inference


21- What does it mean if a Multiple Linear Regression model has a high R² but low adjusted R²?

It means:
Some predictors are not useful
The model includes unnecessary variables

Adjusted R² penalizes useless predictors


22- Why is it important to scale variables in Multiple Linear Regression?

Scaling is important for:

Gradient descent convergence
Regularization (Ridge/Lasso)
Fair coefficient comparison
Numerical stability


23- What is polynomial regression?

Polynomial regression models a non-linear relationship using polynomial terms.

Example:

Y = b0 + b1X + b2X²


24- How does polynomial regression differ from linear regression?

Linear:
Straight line
Y = b0 + b1X

Polynomial:
Curved relationship
Y = b0 + b1X + b2X²


25- When is polynomial regression used?

Used when:

Relationship is curved
Linear model underfits
Trend is non-linear


26- What is the general equation for polynomial regression?

Y = b0 + b1X + b2X² + ... + bnXⁿ


27- Can polynomial regression be applied to multiple variables?

Yes.

Example:

Y = b0 + b1X1² + b2X2² + b3X1X2


28- What are the limitations of polynomial regression?

Limitations:

Overfitting
Poor extrapolation
Sensitive to outliers
High variance
Hard to interpret


29- What methods can be used to evaluate model fit when selecting the degree of a polynomial?

Methods:

Cross-validation
Adjusted R²
RMSE / MSE
AIC / BIC
Validation curve


30- Why is visualization important in polynomial regression?

Visualization helps:

Detect non-linear patterns
Choose correct polynomial degree
See overfitting and underfitting
Interpret model shape clearly
