1. What is the difference between interpreted and compiled languages?
In interpreted languages, code is executed line by line (e.g., Python).  
In compiled languages, the code is converted into machine code before execution (e.g., C, C++).

Example:  
Python → Interpreted  
C → Compiled


2. What is exception handling in Python?
Exception handling is a way to handle runtime errors in a program so that it doesn’t crash.
It uses blocks like try, except, else, and finally.

Example:
try:
    a = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero")


3. What is the purpose of the finally block in exception handling?
The finally block is always executed — whether an exception occurs or not.
It’s commonly used to close files or release resources.

Example:
try:
    f = open("data.txt")
finally:
    f.close()

24. How do you raise an exception manually in Python?
We can raise an exception manually using the raise keyword. 
It helps us stop the program when a specific condition is not met.

Example:
age = int(input("Enter your age: "))
if age < 18:
    raise ValueError("You must be at least 18 years old to continue.")
else:
    print("Access granted!")

Explanation:
The raise statement allows us to trigger an exception ourselves.
Here, if the age is below 18, a ValueError is raised with a custom message.


25. Why is it important to use multithreading in certain applications?
Multithreading allows multiple tasks to run at the same time within one program.
It is useful when one task is waiting (like downloading data), so another task can continue.

Advantages:
- Makes the program faster and more efficient.
- Keeps applications (like GUIs) responsive.
- Good for I/O-bound operations such as file reading or web requests.

Example:
import threading
import time

def display():
    for i in range(5):
        print("Running thread:", i)
        time.sleep(1)

t1 = threading.Thread(target=display)
t2 = threading.Thread(target=display)

t1.start()
t2.start()

t1.join()
t2.join()
print("Both threads completed.")

Explanation:
Here, two threads run the same function at once. 
This allows multiple tasks to execute simultaneously in a single program.

```python
# 22. Check if a file is empty before attempting to read its contents
import os

file_name = "example.txt"

if os.path.exists(file_name):
    if os.stat(file_name).st_size == 0:
        print("The file is empty.")
    else:
        with open(file_name, "r") as f:
            print("File content:\n", f.read())
else:
    print("File does not exist.")
```
