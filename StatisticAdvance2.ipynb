Question 1: What is hypothesis testing in statistics?

Answer:
Hypothesis testing is a statistical method used to make decisions or draw conclusions about a population based on sample data. It helps determine whether there is enough evidence to support a specific claim or assumption about a population parameter such as the mean, proportion, or variance.

In hypothesis testing, we start by stating a null hypothesis and an alternative hypothesis. Then, we collect sample data and perform a statistical test to decide whether to reject the null hypothesis or not. Hypothesis testing is widely used in research, quality control, medicine, and business decision-making.



Question 2: What is the null hypothesis, and how does it differ from the alternative hypothesis?

Answer:
The null hypothesis (H₀) is a statement that assumes there is no effect, no difference, or no relationship in the population. It represents the default or existing condition and is tested against sample data.

The alternative hypothesis (H₁ or Hₐ) is a statement that contradicts the null hypothesis. It suggests that there is an effect, difference, or relationship in the population.

Difference between Null and Alternative Hypothesis:

The null hypothesis states no change or no effect.

The alternative hypothesis states some change or effect exists.

The null hypothesis is tested directly, while the alternative hypothesis is accepted only if the null hypothesis is rejected.



Question 3: Explain the significance level in hypothesis testing and its role in deciding the outcome of a test.

Answer:
The significance level, denoted by α (alpha), is the probability of rejecting the null hypothesis when it is actually true. Common significance levels are 0.05 (5%), 0.01 (1%), and 0.10 (10%).

The significance level sets a threshold for deciding whether the test result is statistically significant. If the p-value obtained from the test is less than or equal to the significance level, the null hypothesis is rejected. If the p-value is greater than the significance level, the null hypothesis is not rejected.

Thus, the significance level plays a crucial role in controlling the risk of making incorrect conclusions in hypothesis testing.



Question 4: What are Type I and Type II errors? Give examples of each.

Answer:
In hypothesis testing, errors can occur when making decisions based on sample data. These errors are called Type I error and Type II error.

A Type I error occurs when the null hypothesis is rejected even though it is actually true. This means we conclude that there is an effect or difference when, in reality, there is none.
Example: A medical test indicates that a person has a disease when the person is actually healthy.

A Type II error occurs when the null hypothesis is not rejected even though it is false. This means we fail to detect an effect or difference that truly exists.
Example: A medical test indicates that a person is healthy when the person actually has the disease.



Question 5: What is the difference between a Z-test and a T-test? Explain when to use each.

Answer:
The Z-test and T-test are statistical tests used to determine whether there is a significant difference between sample data and a population parameter or between two samples.

A Z-test is used when the population standard deviation is known and the sample size is large (usually n ≥ 30). It assumes that the data follows a normal distribution.

A T-test is used when the population standard deviation is unknown and the sample size is small (usually n < 30). It is more flexible and commonly used in real-world situations.

Summary:

Use a Z-test for large samples with known population standard deviation.

Use a T-test for small samples with unknown population standard deviation.

```python
#6: Write a Python program to generate a binomial distribution with n = 10 and p = 0.5, then plot its histogram.
import random
import matplotlib.pyplot as plt

# Parameters
n = 10
p = 0.5
samples = 1000

# Generate binomial distribution
data = []
for _ in range(samples):
    success = 0
    for _ in range(n):
        if random.random() < p:
            success += 1
    data.append(success)

# Plot histogram
plt.hist(data, bins=range(0, n + 2), edgecolor='black')
plt.xlabel("Number of Successes")
plt.ylabel("Frequency")
plt.title("Binomial Distribution (n=10, p=0.5)")
plt.show()
```

```python
#Implement hypothesis testing using Z-statistics for a sample dataset in Python. Show the Python code and interpret the results.

sample_data = [49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6,
               50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,
               50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9,
               50.3, 50.4, 50.0, 49.7, 50.5, 49.9]
import math

# Sample data
data = sample_data
n = len(data)
population_mean = 50
population_std = 1  # assumed known

# Calculate sample mean
sample_mean = sum(data) / n

# Z-statistic
z = (sample_mean - population_mean) / (population_std / math.sqrt(n))

print("Sample Mean:", sample_mean)
print("Z-value:", z)

# Decision
critical_value = 1.96  # for alpha = 0.05 (two-tailed)
if abs(z) > critical_value:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

```python
#8: Write a Python script to simulate data from a normal distribution and calculate the 95% confidence interval for its mean. Plot the data using Matplotlib.
import random
import math
import matplotlib.pyplot as plt

# Generate normal distribution data
n = 100
mean = 50
std_dev = 5

data = [random.gauss(mean, std_dev) for _ in range(n)]

# Calculate sample statistics
sample_mean = sum(data) / n
variance = sum((x - sample_mean) ** 2 for x in data) / (n - 1)
sample_std = math.sqrt(variance)

# 95% confidence interval
z = 1.96
margin_error = z * (sample_std / math.sqrt(n))
lower_bound = sample_mean - margin_error
upper_bound = sample_mean + margin_error

print("Sample Mean:", sample_mean)
print("95% Confidence Interval:", (lower_bound, upper_bound))

# Plot histogram
plt.hist(data, bins=15, edgecolor='black')
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Normal Distribution Data")
plt.show()
```

```python
#9: Write a Python function to calculate the Z-scores from a dataset and visualize the standardized data using a histogram. Explain what the Z-scores represent in terms of standard deviations from the mean.
import math
import matplotlib.pyplot as plt

# Sample dataset
data = [12, 15, 18, 20, 22, 25, 30, 28, 24, 19]

# Function to calculate Z-scores
def calculate_z_scores(data):
    mean = sum(data) / len(data)
    variance = sum((x - mean) ** 2 for x in data) / len(data)
    std_dev = math.sqrt(variance)
    z_scores = [(x - mean) / std_dev for x in data]
    return z_scores

z_scores = calculate_z_scores(data)

print("Z-scores:", z_scores)

# Plot histogram
plt.hist(z_scores, bins=10, edgecolor='black')
plt.xlabel("Z-score")
plt.ylabel("Frequency")
plt.title("Histogram of Z-scores")
plt.show()
```
