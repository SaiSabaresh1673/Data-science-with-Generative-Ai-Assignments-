Question 1: What is the difference between descriptive statistics and inferential statistics? Explain with examples.
Answer:

Descriptive statistics are used to summarize and describe the main features of a dataset. They help in organizing and simplifying large amounts of data using measures like mean, median, mode, tables, charts, and graphs.
Example: Calculating the average marks scored by students in a class.

Inferential statistics are used to make predictions or draw conclusions about a population based on a sample of data. It involves probability, hypothesis testing, confidence intervals, and regression.
Example: Predicting the average marks of all college students by analyzing a sample of students.



Question 2: What is sampling in statistics? Explain the differences between random and stratified sampling.
Answer:

Sampling is the process of selecting a subset of individuals or observations from a population to estimate characteristics of the entire population.

Random Sampling:
Every individual in the population has an equal chance of being selected.
Example: Selecting 50 students randomly from a college.

Stratified Sampling:
The population is divided into different groups (strata) based on characteristics such as age or gender, and samples are taken from each group.
Example: Selecting students separately from each department to ensure fair representation.



Question 3: Define mean, median, and mode. Explain why these measures of central tendency are important.
Answer:

Mean: The average of all values in a dataset.

Median: The middle value when the data is arranged in ascending or descending order.

Mode: The value that occurs most frequently.

These measures of central tendency help summarize data with a single representative value, making it easier to understand and compare datasets.



Question 4: Explain skewness and kurtosis. What does a positive skew imply about the data?
Answer:

Skewness measures the asymmetry of a data distribution.

Positive skew: Tail is longer on the right side

Negative skew: Tail is longer on the left side

Kurtosis measures the peakedness or flatness of a distribution.

A positive skew indicates that most values are concentrated on the left and a few large values stretch the distribution to the right.

Question 5: Implement a Python program to compute the mean, median, and mode of a given list of numbers.

numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]

```python
import statistics

numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]

print("Mean:", statistics.mean(numbers))
print("Median:", statistics.median(numbers))
print("Mode:", statistics.mode(numbers))
```

Question 6: Compute the covariance and correlation coefficient between the following two datasets.

list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]

```python
import numpy as np

list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]

covariance = np.cov(list_x, list_y)[0][1]
correlation = np.corrcoef(list_x, list_y)[0][1]

print("Covariance:", covariance)
print("Correlation:", correlation)
```

Question 7: Write a Python script to draw a boxplot and identify outliers. Explain the result.

data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]

The boxplot displays the median, quartiles, and spread of the data. The value 35 appears as an outlier because it is far away from the rest of the values.

```python
import matplotlib.pyplot as plt

data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]

plt.boxplot(data)
plt.show()
```

Question 8: Relationship between advertising spend and daily sales
Answer:

Covariance indicates the direction of the relationship, while correlation measures the strength of the relationship. A positive correlation means that as advertising spend increases, daily sales also increase.

```python
import numpy as np

advertising_spend = [200, 250, 300, 400, 500]
daily_sales = [2200, 2450, 2750, 3200, 4000]

correlation = np.corrcoef(advertising_spend, daily_sales)[0][1]
print("Correlation:", correlation)
```
