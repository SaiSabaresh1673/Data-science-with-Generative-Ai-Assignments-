                                         Therotical questions
1. What are data structures, and why are they important?  
ANS-  Data structures are ways to organize and store data efficiently so that it can be accessed and modified easily. Examples are lists, tuples, sets, and dictionaries. They are important because they make operations like searching, sorting, and managing data faster and easier.  

2. Explain the difference between mutable and immutable data types with examples  
ANS- Mutable data types can be changed after creation, such as lists and dictionaries. Example:
list1 = [1, 2, 3]
list1[0] = 10
Immutable data types cannot be changed after creation, such as strings and tuples. Example:
text = "hello"
text[0] = "H" # gives an error  

3. What are the main differences between lists and tuples in Python  
ANS-Lists are mutable, written with square brackets, and can be modified.
Tuples are immutable, written with parentheses, and cannot be modified.
Lists are slower and use more memory, while tuples are faster and use less memory.

4. Describe how dictionaries store data  
ANS- Dictionaries store data as key-value pairs. Each key is unique and is used to access its value quickly. Example:
student = {"name": "Sabaresh", "age": 21}

5. Why might you use a set instead of a list in Python  
ANS- A set is used when you want to store unique values and perform fast membership checks. Sets automatically remove duplicate values.

6. What is a string in Python, and how is it different from a list  
ANS- A string is a sequence of characters and is immutable, while a list can store elements of different data types and is mutable.

7. How do tuples ensure data integrity in Python  
ANS- Tuples are immutable, meaning their values cannot be changed after creation. This ensures that the data remains constant and secure from modification.

8. What is a hash table, and how does it relate to dictionaries in Python  
ANS- A hash table stores data using a hash function that assigns keys to specific memory locations. Dictionaries in Python are implemented using hash tables, allowing fast access to values using keys.

9. Can lists contain different data types in Python  
ANS- Yes, lists can store elements of different data types. Example: [1 , "hello", 3.5, True]

10. Explain why strings are immutable in Python  
ANS- Strings are immutable to make them more secure and efficient. When a string is modified, a new string is created instead of changing the original one.

11. What advantages do dictionaries offer over lists for certain tasks    
ANS- Dictionaries allow quick access to values using keys instead of indexes and make the data more readable and structured. They are faster for lookups and updates.

12. Describe a scenario where using a tuple would be preferable over a list  
ANS- Tuples are preferable when the data should not change, such as storing coordinates (x, y) or fixed configuration values.

13. How do sets handle duplicate values in Python  
ANS- Sets automatically remove duplicate values. Example: {1, 2, 2, 3} becomes {1, 2, 3}

14. How does the “in” keyword work differently for lists and dictionaries  
ANS- In a list, it checks if a value exists in the list.
      In a dictionary, it checks if a key exists in the dictionary.

15. Can you modify the elements of a tuple? Explain why or why not  
ANS- No, tuples are immutable, so once created, their elements cannot be changed, added, or removed.

16. What is a nested dictionary, and give an example of its use case  
ANS- A nested dictionary is a dictionary within another dictionary. It is used to store structured data. Example:
students = {"Sabaresh": {"age": 21, "branch": "CSE"}, "Vigash": {"age": 22, "branch": "ECE"}}

17. Describe the time complexity of accessing elements in a dictionary  
ANS- The average time complexity is O(1), meaning it is very fast. In rare worst cases, it can be O(n).

18. In what situations are lists preferred over dictionaries  
ANS- Lists are preferred when the order of elements matters or when you only need to store a collection of items without any key-value mapping.

19. Why are dictionaries considered unordered, and how does that affect data retrieval  
ANS- Dictionaries are considered unordered because they store elements based on hash values of keys, not by position. Although Python 3.7+ preserves insertion order, logically, dictionaries are treated as unordered.

20. Explain the difference between a list and a dictionary in terms of data retrieval  
ANS- Lists are accessed by their index position, such as mylist[0].
     Dictionaries are accessed by their key, such as mydict["name"].
     Dictionaries provide faster lookups compared to lists.
