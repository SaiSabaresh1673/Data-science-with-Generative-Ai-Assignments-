1. What is the difference between a function and a method in Python?

Function: Defined using the def keyword and can be called independently.
Method: A function that belongs to an object (called using dot notation).

Example:
def greet():
    print("Hello!")   # Function

name = "Python"
print(name.upper())   # upper() is a method

2. Explain the concept of function arguments and parameters in Python.

Parameters are variables listed inside parentheses in the function definition.
Arguments are values passed to those parameters when the function is called.

Example:
def add(a, b):   # a, b are parameters
    return a + b

print(add(3, 5))  # 3, 5 are arguments

3. What are the different ways to define and call a function in Python?

You can define functions using:
1. def keyword
2. lambda expressions (anonymous functions)

Example:
def square(x):
    return x * x

sq = lambda x: x * x  # Lambda way

print(square(4))
print(sq(5))

4. What is the purpose of the return statement in a Python function?

The return statement sends a value back to the caller from a function.
Without it, the function returns None by default.

Example:
def multiply(a, b):
    return a * b

result = multiply(2, 4)
print(result)

5. What are iterators in Python and how do they differ from iterables?

Iterable: Any object that can return an iterator (like lists, tuples, strings).
Iterator: An object that returns items one at a time using the __next__() method.

Example:
nums = [1, 2, 3]
it = iter(nums)  # Create iterator

print(next(it))
print(next(it))

6. Explain the concept of generators in Python and how they are defined.

Generators are special functions that yield values one at a time using the yield keyword.
They generate items lazily, saving memory.

Example:
def count_up_to(n):
    for i in range(1, n+1):
        yield i

for num in count_up_to(3):
    print(num)

7. What are the advantages of using generators over regular functions?

1. Save memory (lazy evaluation)
2. Useful for large datasets or streams
3. Maintain state automatically between yields

Example:
def squares():
    for i in range(5):
        yield i * i

print(list(squares()))

8. What is a lambda function in Python and when is it typically used?

A lambda function is a small anonymous function defined with the lambda keyword.
Used for short, temporary functions without using def.

Example:
double = lambda x: x * 2
print(double(5))

9. Explain the purpose and usage of the map() function in Python.

map() applies a given function to all items in an iterable.
It returns a map object (iterator).

Example:
nums = [1, 2, 3, 4]
squares = map(lambda x: x**2, nums)
print(list(squares))

10. What is the difference between map(), reduce(), and filter() functions in Python?

Function | Purpose | Returns
map()    | Applies a function to all elements | Map object
filter() | Selects elements that satisfy a condition | Filter object
reduce() | Reduces elements to a single value | Single value

Example:
from functools import reduce

nums = [1, 2, 3, 4]

mapped = list(map(lambda x: x*2, nums))
filtered = list(filter(lambda x: x%2==0, nums))
reduced = reduce(lambda a, b: a + b, nums)

print(mapped, filtered, reduced)

```python
# 1. Write a Python function that takes a list of numbers as input 
# and returns the sum of all even numbers in the list.

def sum_even_numbers(numbers):
    return sum(num for num in numbers if num % 2 == 0)

nums = [1, 2, 3, 4, 5, 6]
print("Sum of even numbers:", sum_even_numbers(nums))  
```

```python
# 2. Create a Python function that accepts a string and returns the reverse of that string.

def reverse_string(s):
    return s[::-1]

text = "Python"
print("Reversed string:", reverse_string(text))  
```

```python
# 3. Implement a Python function that takes a list of integers 
# and returns a new list containing the squares of each number.

def square_list(numbers):
    return [n ** 2 for n in numbers]

nums = [1, 2, 3, 4, 5]
print("Squared list:", square_list(nums))
```

```python
# 4. Write a Python function that checks if a given number is prime or not from 1 to 200.

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

for num in range(1, 201):
    if is_prime(num):
        print(num, end=" ")
```

```python
# 5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.

def fib(n):
    a=0
    b=1
    for i in range(n):
        yield a
        a,b=b,a+b

f = fib(10)
for num in f:
    print(num, end=" ")
```

```python
# 6. Write a generator function in Python that yields the powers of 2 up to a given exponent.

def powers_of_two(n):
    for i in range(n+1):
        yield 2 ** i

for p in powers_of_two(5):
    print(p, end=" ")
```

```python
# 8. Use a lambda function in Python to sort a list of tuples based on the second element of each tuple.

tuples_list = [(1, 3), (4, 1), (2, 5)]
sorted_list = sorted(tuples_list, key=lambda x: x[1])
print(sorted_list) 
```

```python
# 9. Write a Python program that uses map() to convert a list of temperatures from Celsius to Fahrenheit.

celsius = [0, 20, 37, 100]
fahrenheit = list(map(lambda c: (c * 9/5) + 32, celsius))
print(fahrenheit) 
```

```python
# 10. Create a Python program that uses filter() to remove all the vowels from a given string.

text = "Hello World"
no_vowels = ''.join(filter(lambda x: x.lower() not in 'aeiou', text))
print(no_vowels) 
```
