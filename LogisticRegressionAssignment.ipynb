Question 1: What is Simple Linear Regression (SLR)? Explain its purpose

Simple Linear Regression (SLR) is a statistical technique used to model the
relationship between one independent variable (X) and one dependent variable (Y)
by fitting a straight line to the observed data.

It predicts how the dependent variable changes when the independent variable changes.

Purpose:
1. Prediction – Estimate unknown or future values
2. Understanding relationship between variables
3. Measuring impact of X on Y
4. Identifying trends in data


---------------------------------------------------------------------

Question 2: What are the key assumptions of Simple Linear Regression?

1. Linearity
   The relationship between X and Y must be linear (straight line relationship).

2. Independence of Errors
   Residuals (errors) must be independent of each other.

3. Homoscedasticity
   Variance of errors must remain constant for all values of X.

4. Normality of Errors
   Errors should follow normal distribution.

5. No Measurement Error in Independent Variable
   X values should be measured accurately.

Question 3: Write the mathematical equation for a simple linear regression model
and explain each term.

Mathematical Model:

Y = β0 + β1X + ε

Where:
Y   = Dependent variable (output/target)
X   = Independent variable (input/predictor)
β0  = Intercept (value of Y when X = 0)
β1  = Slope (change in Y for 1 unit change in X)
ε   = Error term (difference between actual and predicted value)

Predicted Equation:
Ŷ = β0 + β1X

Question 5: What is the method of least squares in linear regression?

The Method of Least Squares is used to find the best fitting regression line.

It selects β0 and β1 such that the sum of squared errors is minimum.

Error:
ei = Yi - Ŷi

Objective Function:
Minimize  Σ(Yi - Ŷi)²
         = Σ(Yi - (β0 + β1Xi))²

Slope:
β1 = Σ[(Xi - X̄)(Yi - Ȳ)] / Σ[(Xi - X̄)²]

Intercept:
β0 = Ȳ - β1X̄

This method gives the best straight line that fits the data.

Question 6: What is Logistic Regression? How does it differ from Linear Regression?

Logistic Regression is a supervised machine learning algorithm used for
classification problems. It predicts the probability that an observation belongs
to a particular category (usually 0 or 1).

It uses a sigmoid function to convert linear output into probability:

P(Y=1) = 1 / (1 + e^-(β0 + β1X))

Difference between Linear Regression and Logistic Regression:

Linear Regression:
- Used for predicting continuous values
- Output range: (-∞ to +∞)
- Uses straight line equation Y = β0 + β1X
- Example: Predict salary, temperature, price

Logistic Regression:
- Used for classification problems
- Output range: 0 to 1 (probability)
- Uses sigmoid (S-shaped) curve
- Example: Spam detection, disease prediction

Question 7: Name and briefly describe three common evaluation metrics for regression models.

1. Mean Absolute Error (MAE)
   Average of absolute differences between actual and predicted values.
   MAE = (1/n) Σ |Yi - Ŷi|
   Lower value indicates better model.

2. Mean Squared Error (MSE)
   Average of squared differences between actual and predicted values.
   MSE = (1/n) Σ (Yi - Ŷi)²
   Penalizes large errors more strongly.

3. Root Mean Squared Error (RMSE)
   Square root of MSE.
   RMSE = √MSE
   Gives error in same units as output variable.

Question 8: What is the purpose of the R-squared metric in regression analysis?

R-squared (Coefficient of Determination) measures how well the regression
model explains the variability of the dependent variable.

Formula:
R² = 1 - (SS_res / SS_tot)

Where:
SS_res = Sum of squared residuals
SS_tot = Total sum of squares

Interpretation:
R² = 0   → Model explains none of the variance
R² = 1   → Model explains all of the variance

Higher R² means better fit of the model.

Question 10: How do you interpret the coefficients in a simple linear regression model?

In equation:
Y = β0 + β1X

β0 (Intercept):
Represents the predicted value of Y when X = 0.

β1 (Slope):
Represents how much Y changes when X increases by 1 unit.

Example:
If equation is:
Salary = 15000 + 3000 × Experience

Interpretation:
Intercept (15000) → Base salary with 0 years experience
Slope (3000) → Salary increases by ₹3000 for each additional year

```python
#Question 9: Write Python code to fit a simple linear regression model using scikit-learn
#and print the slope and intercept.

import numpy as np
from sklearn.linear_model import LinearRegression

# Sample data
X = np.array([1,2,3,4,5]).reshape(-1,1)
Y = np.array([2,4,6,8,10])

# Create model
model = LinearRegression()

# Train model
model.fit(X, Y)

# Print slope and intercept
print("Slope (β1):", model.coef_[0])
print("Intercept (β0):", model.intercept_)
```
